```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =========================================================
# Feature Engineering
# =========================================================

df['GPA_Change'] = (
    df['Post_Semester_GPA']
    - df['Pre_Semester_GPA']
)

df['AI_Usage_Category'] = pd.cut(
    df['Weekly_GenAI_Hours'],
    bins=[0,5,15,30,100],
    labels=['Low','Moderate','High','Extreme']
)

# =========================================================
# KPI Values
# =========================================================

avg_ai = round(df['Weekly_GenAI_Hours'].mean(),2)

avg_gpa_change = round(
    df['GPA_Change'].mean(),2
)

avg_dependency = round(
    df['Perceived_AI_Dependency'].mean(),2
)

avg_retention = round(
    df['Skill_Retention_Score'].mean(),2
)

# =========================================================
# Aggregations
# =========================================================

major_ai = df.groupby(
    'Major_Category'
)['Weekly_GenAI_Hours'].mean().reset_index()

use_case = df['Primary_Use_Case'].value_counts().reset_index()

use_case.columns = [
    'Primary_Use_Case',
    'Count'
]

paid_gpa = df.groupby(
    'Paid_Subscription'
)['GPA_Change'].mean().reset_index()

prompt_retention = df.groupby(
    'Prompt_Engineering_Skill'
)['Skill_Retention_Score'].mean().reset_index()

burnout_dependency = df.groupby(
    'Burnout_Risk_Level'
)['Perceived_AI_Dependency'].mean().reset_index()


# =========================================================
# Dashboard Layout
# =========================================================

fig = make_subplots(

rows=4,
cols=4,

subplot_titles=(

'','','','',

'AI Usage by Major',
'Pre vs Post GPA',
'GPA Change Distribution',
'AI Usage vs Traditional Study',

'Primary AI Use Cases',
'Burnout vs Dependency',
'Paid vs Free GPA Change',
'Skill Retention Distribution',
'','','','','',''

),

specs=[

[{"type":"indicator"},
 {"type":"indicator"},
 {"type":"indicator"},
 {"type":"indicator"}],

[
{"type":"bar"},
{"type":"bar"},
{"type":"histogram"},
{"type":"scatter"}
],

[
{"type":"pie"},
{"type":"bar"},
{"type":"bar"},
{"type":"histogram"}
],

[None,None,None,None]

],

vertical_spacing=0.12,
horizontal_spacing=0.08

)

# =========================================================
# KPI CARDS
# =========================================================

fig.add_trace(

go.Indicator(
mode="number",
value=avg_ai,
title={'text':
"<b>Avg Weekly AI Hours</b>"}
),

row=1,col=1

)

fig.add_trace(

go.Indicator(
mode="number",
value=avg_gpa_change,
title={'text':
"<b>Avg GPA Change</b>"}
),

row=1,col=2

)

fig.add_trace(

go.Indicator(
mode="number",
value=avg_dependency,
title={'text':
"<b>Avg AI Dependency</b>"}
),

row=1,col=3

)

fig.add_trace(

go.Indicator(
mode="number",
value=avg_retention,
title={'text':
"<b>Avg Skill Retention</b>"}
),

row=1,col=4

)

# =========================================================
# AI Usage by Major
# =========================================================

fig.add_trace(

go.Bar(
x=major_ai['Major_Category'],
y=major_ai['Weekly_GenAI_Hours'],
name='AI Usage'
),

row=2,
col=1

)

# ==============================
# GPA Comparison
# ==============================

fig.add_trace(

go.Bar(

x=['Pre GPA','Post GPA'],

y=[

df['Pre_Semester_GPA'].mean(),
df['Post_Semester_GPA'].mean()

]

),

row=2,
col=2

)


# =========================================================
# GPA Change Distribution
# =========================================================

fig.add_trace(

go.Histogram(
x=df['GPA_Change'],
nbinsx=30,
name='GPA'
),

row=2,
col=3

)

# =========================================================
# AI Hours vs Study Hours
# =========================================================

fig.add_trace(

go.Scatter(
x=df['Weekly_GenAI_Hours'],
y=df['Traditional_Study_Hours'],
mode='markers'
),

row=2,
col=4

)

# =========================================================
# Primary Use Cases
# =========================================================

fig.add_trace(

go.Pie(

labels=use_case['Primary_Use_Case'],
values=use_case['Count'],

textinfo='label+percent',
hoverinfo='label+value+percent',
showlegend=True

),

row=3,
col=1

)

# =========================================================
# Burnout vs Dependency
# =========================================================

fig.add_trace(

go.Bar(

x=burnout_dependency[
'Burnout_Risk_Level'
],

y=burnout_dependency[
'Perceived_AI_Dependency'
]

),

row=3,
col=2

)

# ==============================
# Paid vs Free GPA
# ==============================

fig.add_trace(

go.Bar(

x=paid_gpa[
'Paid_Subscription'
],

y=paid_gpa[
'GPA_Change'
]

),

row=3,
col=3

)



# =========================================================
# Skill Retention
# =========================================================

fig.add_trace(

go.Histogram(
x=df['Skill_Retention_Score'],
nbinsx=30
),

row=3,
col=4

)

# =========================================================
# Layout
# =========================================================

fig.update_layout(

height=1100,
width=2000,

title={

'text':
"<b>Impact of AI on Students Dashboard</b><br><sup>AI Usage → Study Habits → Academic Performance → Mental Health</sup>",

'x':0.5,
'xanchor':'center'

},

template='plotly_white',

showlegend=False,

font=dict(size=11)

)

# =========================================================
# AXIS TITLES
# =========================================================

# Row 2

# AI Usage by Major
fig.update_xaxes(
    title_text="Major Category",
    row=2,
    col=1
)

fig.update_yaxes(
    title_text="Average Weekly AI Hours",
    row=2,
    col=1
)

# Pre vs Post GPA
fig.update_xaxes(
    title_text="Semester Type",
    row=2,
    col=2
)

fig.update_yaxes(
    title_text="Average GPA",
    row=2,
    col=2
)

# GPA Change Distribution
fig.update_xaxes(
    title_text="GPA Change",
    row=2,
    col=3
)

fig.update_yaxes(
    title_text="Number of Students",
    row=2,
    col=3
)

# AI vs Traditional Study
fig.update_xaxes(
    title_text="Weekly AI Hours",
    row=2,
    col=4
)

fig.update_yaxes(
    title_text="Traditional Study Hours",
    row=2,
    col=4
)

# Row 3

# Burnout vs Dependency
fig.update_xaxes(
    title_text="Burnout Risk Level",
    row=3,
    col=2
)

fig.update_yaxes(
    title_text="Average AI Dependency Score",
    row=3,
    col=2
)

# Paid vs Free GPA Change
fig.update_xaxes(
    title_text="Subscription Type",
    row=3,
    col=3
)

fig.update_yaxes(
    title_text="Average GPA Change",
    row=3,
    col=3
)

# Skill Retention Distribution
fig.update_xaxes(
    title_text="Skill Retention Score",
    row=3,
    col=4
)

fig.update_yaxes(
    title_text="Number of Students",
    row=3,
    col=4
)

fig.show()
```

## Key Insights

1. Students spend an average of 8.43 hours per week using AI tools, showing that AI has become a regular component of academic activities rather than an occasional resource.
2. The average GPA change is +0.20, indicating a small but positive improvement in academic performance after AI adoption.
3. The average AI dependency score is 3.51, suggesting students rely on AI moderately for learning and academic support.
4. Students maintain a high average skill retention score of 75.8, indicating that AI usage does not appear to significantly reduce knowledge retention.
5. STEM students demonstrate the highest AI usage among all major categories, suggesting stronger AI integration in technical disciplines.
6. The post-semester GPA is higher than the pre-semester GPA, suggesting that AI tools may positively contribute to academic performance.
7. The GPA change distribution is centered slightly toward positive values, indicating that most students experienced small academic improvements rather than dramatic changes.
8. The scatter plot shows a downward pattern, suggesting students with higher AI usage tend to spend fewer hours in traditional study methods.
9. Debugging/Troubleshooting (24.6%) and Copywriting/Drafting (24%) are the dominant use cases, showing students primarily use AI for productivity and assistance tasks.
10. Students with high burnout risk demonstrate greater AI dependency compared to low-risk students, indicating a possible relationship between reliance and mental strain.
11. Students using paid subscriptions show only a slightly higher GPA improvement than free users, suggesting that access alone may not strongly influence academic outcomes.
12. Skill retention scores are concentrated between 70–90, indicating that the majority of students continue to retain knowledge effectively.
